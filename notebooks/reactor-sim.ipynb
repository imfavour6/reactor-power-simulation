##  Constant Reactivity

```python
#Example for understanding 
#Reactivity = How much the reactor wants to change power
reactivity  = - 0.2

if reactivity > 0:
    print("Positive reactivity, Power Increasing ")
elif reactivity < 0:
    print("Negative reactivity, Power Decreasing ")
else:
    print("Zero reactivity, Power Constant ")
```

```python
#starting neutrons in the reactor
neutrons =  1000

# reactivity
reactivity = 0.02

#Create lists to store the history of neutrons and time
neutron_history = []
time_history = []

#Append the initial number of neutrons and time to their respective histories
neutron_history.append(neutrons)  # Append the initial number of neutrons to the neutron_history
time_history.append(0)  # Append the initial time to the time_history

# Time settings
dt = 0.5
number_of_steps = 50
print("Time:", 0, "seconds", "Neutrons:", neutrons)


for step in range(1, number_of_steps):
    time = step * dt  # Calculate the current time based on the step and time increment
    
    change_in_neutrons = neutrons * reactivity # Calculate the change in neutrons based on reactivity
    new_neutrons = neutrons + change_in_neutrons # Calculate the new number of neutrons after the change
    neutrons = new_neutrons  # Update neutrons for the next iteration
    neutron_history.append(neutrons)  # Append the new number of neutrons to the neutron_history
    time_history.append(time)  # Append the current time to the time_history
    print('Time:', time, 'seconds, ' 'Neutrons:', neutrons)  # Print the time and the new number of neutrons

print("Neutron History:", neutron_history)  # Print the history of neutrons
print("Time History:", time_history)  # Print the history of time
```

```python
import pandas as pd

data = pd.DataFrame({'Time(s)': time_history, 'Neutron Population': neutron_history})  # Create a DataFrame with time and neutron history
data.round(2).head(10) # Round the values in the DataFrame to 2 decimal places

data.to_csv("../data/simulation_results.csv", index=False)
```

```python
import matplotlib.pyplot as plt

plt.plot(data['Time(s)'], data['Neutron Population'], marker="o")  # Plot the neutron population against time

plt.xlabel('Time (s)')  # Label the x-axis as 'Time (s)'
plt.ylabel('Neutron Population')  # Label the y-axis as 'Neutron Population'

plt.title('Neutron Population vs Time')  # Set the title of the plot
plt.grid()  # Enable grid lines on the plot
plt.legend(['Neutron Population'])

plt.show()  # Display the plot

plt.savefig("../images/neutron_population.png", dpi=300, bbox_inches="tight")
```

##  Time Dependent Reactivity

```python
# Initial number of neutrons in the reactor
neutrons = 1000

# Time settings
dt = 0.5
number_of_steps = 50

#Create lists to store the history of neutrons and time
neutron_history = []
reactivity_history = []
time_history = []

#Time loop to simulate the change in neutrons over time
for step in range(1, number_of_steps):
    time = step * dt
    if time < 10:
        reactivity = 0
    else:
        reactivity = 0.02
    change_in_neutrons = neutrons * reactivity
    new_neutrons = neutrons + change_in_neutrons
    neutrons = new_neutrons
    neutron_history.append(neutrons)
    reactivity_history.append(reactivity)
    time_history.append(time)
    print ('Time:', time, 'Reactivity:', reactivity, 'Neutrons:', neutrons)
    
```

```python
import pandas as pd
df = pd.DataFrame({'Time(s)': time_history, 'Reactivity': reactivity_history, 'Neutron Population': neutron_history})
df.head(30)
```

```python
import matplotlib.pyplot as plt

plt.plot(df['Time(s)'], df['Neutron Population'], marker="o")  # Plot the neutron population against time

plt.xlabel('Time (s)')  # Label the x-axis as 'Time (s)'
plt.ylabel('Neutron Population')  # Label the y-axis as 'Neutron Population'

plt.title('Neutron Population vs Time')  # Set the title of the plot
plt.grid()  # Enable grid lines on the plot
plt.legend(['Neutron Population'])

plt.show()  # Display the plot
```
