This project investigates reactor power behavior using computational modeling in Python. The simulation explores the effects of reactivity changes, control rod movement, and temperature feedback on reactor power over time.

##  Constant Reactivity

```python
#Example for understanding 
#Reactivity = How much the reactor wants to change power
reactivity  = - 0.2

if reactivity > 0:
    print("Positive reactivity, Power Increasing ")
elif reactivity < 0:
    print("Negative reactivity, Power Decreasing ")
else:
    print("Zero reactivity, Power Constant ")
```

```python
#starting neutrons in the reactor
neutrons =  1000

# reactivity
reactivity = 0.02

#Create lists to store the history of neutrons and time
neutron_history = []
time_history = []

#Append the initial number of neutrons and time to their respective histories
neutron_history.append(neutrons)  # Append the initial number of neutrons to the neutron_history
time_history.append(0)  # Append the initial time to the time_history

# Time settings
dt = 0.5
number_of_steps = 50
print("Time:", 0, "seconds", "Neutrons:", neutrons)


for step in range(1, number_of_steps):
    time = step * dt  # Calculate the current time based on the step and time increment
    
    change_in_neutrons = neutrons * reactivity * dt# Calculate the change in neutrons based on reactivity
    new_neutrons = neutrons + change_in_neutrons # Calculate the new number of neutrons after the change
    neutrons = new_neutrons  # Update neutrons for the next iteration
    neutron_history.append(neutrons)  # Append the new number of neutrons to the neutron_history
    time_history.append(time)  # Append the current time to the time_history
    print('Time:', time, 'seconds, ' 'Neutrons:', neutrons)  # Print the time and the new number of neutrons

print("Neutron History:", neutron_history)  # Print the history of neutrons
print("Time History:", time_history)  # Print the history of time
```

```python
import pandas as pd

data = pd.DataFrame({'Time(s)': time_history, 'Neutron Population': neutron_history})  # Create a DataFrame with time and neutron history
data.round(2).head(10) # Round the values in the DataFrame to 2 decimal places

data.to_csv("../data/constant_reactivity_simulation.csv", index=False)
```

## Control Rods


```python
rod_position = 0.2  # Initial position of the control rod (0.0 = fully withdrawn, 1.0 = fully inserted)
max_rod_reactivity = -0.03 # Reactivity effect when rods are fully inserted

base_reactivity = 0.02  # Base reactivity of the reactor without control rod insertion
neutrons = 1000  # Initial number of neutrons in the reactor

# Time settings
dt = 0.5
number_of_steps = 50

rod_history = []  # List to store the history of control rod positions
neutron_history = []  # List to store the history of neutron populations
total_reactivity_history = []  # List to store the history of total reactivity values
time_history = []  # List to store the history of time values
for step in range(1, number_of_steps):
    time = step * dt
    if time < 10:
        rod_position = 0.2
    elif time < 20:
        rod_position = 0.5
    else:
        rod_position = 0.8
    rod_reactivity = rod_position * max_rod_reactivity  # Calculate the reactivity based on the control rod position
    total_reactivity = base_reactivity + rod_reactivity  # Calculate the total reactivity considering the control rod effect
    change_in_neutrons = neutrons * total_reactivity * dt  # Calculate the change in neutrons based on total reactivity
    new_neutrons = neutrons + change_in_neutrons  # Calculate the new number of neutrons after the change
    neutrons = new_neutrons  # Update neutrons for the next iteration


    rod_history.append(rod_position)  # Append the current control rod position to the rod_history
    neutron_history.append(neutrons)  # Append the current number of neutrons to the neutron_history
    total_reactivity_history.append(total_reactivity)  # Append the current total reactivity to the total_reactivity_history
    time_history.append(time)  # Append the current time to the time_history
    print('Time:', time, 'seconds, ' 'Rod Position:', rod_position, 'Total Reactivity:', total_reactivity, 'Neutrons:', neutrons)  # Print the time, control rod position, total reactivity, and number of neutrons
```

## Temperature Feedback

```python
# Initial reactor conditions
initial_neutrons = 1000
neutrons = initial_neutrons

initial_temperature = 300
temperature = initial_temperature

# Effective growth-rate parameters
base_reactivity = 0.02
rod_position = 0.2
max_rod_reactivity = -0.03

# Temperature feedback coefficient
alpha = 0.004

# Simplified thermal model parameters
heating_rate = 0.2
cooling_coefficient = 0.04

# Time settings
dt = 0.5
number_of_steps = 50
```

```python
time_history = []
neutron_history = []
temperature_history = []
power_history = []

rod_history = []
temperature_feedback_history = []
total_reactivity_history = []
```

```python
for step in range(number_of_steps):

    # Current simulation time
    time = step * dt

    # Calculate normalized reactor power
    normalized_power = neutrons / initial_neutrons

    # Calculate control rod contribution
    rod_reactivity = (
        rod_position * max_rod_reactivity
    )

    # Calculate negative temperature feedback
    temperature_feedback = (
        -alpha * (temperature - initial_temperature)
    )

    # Calculate total effective growth rate
    total_reactivity = (
        base_reactivity
        + rod_reactivity
        + temperature_feedback
    )

    # Store current conditions
    time_history.append(time)
    neutron_history.append(neutrons)
    temperature_history.append(temperature)
    power_history.append(normalized_power)

    rod_history.append(rod_position)
    temperature_feedback_history.append(
        temperature_feedback
    )

    total_reactivity_history.append(
        total_reactivity
    )

    # Calculate change in neutron population
    change_in_neutrons = (
        neutrons * total_reactivity * dt
    )

    # Calculate change in temperature
    change_in_temperature = (
        heating_rate * normalized_power
        - cooling_coefficient
        * (temperature - initial_temperature)
    ) * dt

    # Update the reactor state
    neutrons = neutrons + change_in_neutrons

    temperature = temperature + change_in_temperature
```

```python
temperature_df = pd.DataFrame({

    'Time(s)': time_history,

    'Neutron Population': neutron_history,

    'Normalized Power': power_history,

    'Temperature(C)': temperature_history,

    'Rod Position': rod_history,

    'Temperature Feedback': temperature_feedback_history,

    'Total Reactivity': total_reactivity_history

})

temperature_df.head(10)
temperature_df.to_csv(
    "../data/temperature_feedback_simulation.csv",
    index=False
)
```

## Comparison of Reactor Power With and Without Temperature Feedback

```python
# Initial conditions
neutrons_no_feedback = initial_neutrons

# Same initial growth rate as the feedback simulation
constant_growth_rate = (
    base_reactivity
    + rod_position * max_rod_reactivity
)

# Store the results
no_feedback_power = []
no_feedback_time = []

for step in range(number_of_steps):

    time = step * dt

    # Calculate normalized power
    normalized_power = (
        neutrons_no_feedback / initial_neutrons
    )

    # Store current values
    no_feedback_time.append(time)
    no_feedback_power.append(normalized_power)

    # Calculate change in neutrons
    change_in_neutrons = (
        neutrons_no_feedback
        * constant_growth_rate
        * dt
    )

    # Update neutron population
    neutrons_no_feedback += change_in_neutrons
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    no_feedback_time,
    no_feedback_power,
    label="Without Temperature Feedback",
    linewidth=2
)

plt.plot(
    temperature_df['Time(s)'],
    temperature_df['Normalized Power'],
    label="With Temperature Feedback",
    linewidth=2
)

plt.axhline(
    1.0,
    color="black",
    linestyle="--",
    label="Initial Power"
)

plt.xlabel("Time (s)")
plt.ylabel("Normalized Reactor Power")

plt.title(
    "Comparison of Reactor Power With and Without Temperature Feedback"
)

plt.grid(True)
plt.legend()

plt.savefig(
    "../images/temperature_feedback_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## Results and Discussion

### Constant Reactivity

The constant reactivity simulation demonstrates how neutron population changes when the effective growth rate remains fixed. A positive growth rate produces an increasing neutron population, while a negative growth rate produces a decreasing population.

### Time-Dependent Reactivity

The time-dependent simulation introduces changes in the effective growth rate at specified time intervals. This allows the model to demonstrate stable, decreasing, and increasing neutron population behavior within the same simulation.

### Control Rod Movement

The control rod simulation models the effect of increasing rod insertion on neutron population and normalized reactor power. Greater rod insertion introduces a more negative growth-rate contribution, reducing the neutron population growth rate or causing it to decrease.

### Temperature Feedback

The temperature feedback simulation introduces a negative feedback mechanism in which increasing temperature reduces the effective growth rate. This demonstrates how temperature-dependent feedback can oppose an increase in reactor power.

### Comparison of Reactor Power

The comparison between simulations with and without temperature feedback illustrates the influence of negative feedback on power behavior. Both simulations begin with identical initial conditions, allowing their different responses to be attributed to the temperature feedback model.

### Model Limitations

This project uses simplified reactor dynamics to investigate qualitative relationships between neutron population, reactor power, reactivity, control rod insertion, and temperature feedback. The numerical parameters are illustrative rather than representative of a specific nuclear reactor.

The model does not include delayed neutron precursor groups, neutron generation time, spatial neutron flux distributions, detailed thermal hydraulics, or nonlinear control rod worth. The variable labeled reactivity is implemented as an effective neutron population growth rate rather than standard dimensionless reactor reactivity.

Consequently, the simulations are intended for educational analysis and should not be used to predict actual nuclear reactor operating behavior or safety margins.
